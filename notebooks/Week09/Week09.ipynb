### dropping columns
The dataset might not have columns that are relevant to the analysis, so to get rid of them, we can use `df.drop()`.
We add the column name and `axis=1`, which means we're dropping a column, rather than a row.

```python
df = df.drop('Unnamed: 0', axis=1) # this returns a new dataframe. You can make a new variable, or re-define df so you don't get a bunch of dfs.
```

### Filtering data
We can filter dfs using comparison operators like `==`. Comparing a column to a value will return a Series of every row in the columns, and whether that matches teh criteria. 

```python
df['title'] == "Woman" # will give 500 responses for the 500 rows/itmes
```

```python
woman_mask = df['title'] == "Woman" #store the series in a new var called woman_mask
```

```python
df[woman_mask][0:3] # if we use the series as the querying term, it will show us only where the series value is True!
```

woman_objects = df[woman_mask]

We can aslo use `df[column].str.contains('search term', na=False)`. Contains will help us find inexact matches.

We need to include `na=False` to handle missing values, otherwise the mask will inherit those incorrectly.

```python
courbet = df['artistDisplayName'].str.contains('Courbet', na=False)
df[courbet] # using the above mask to display the dataframe
```

### Graphing data

Pandas can generate graphs and charts using `.plot()`. We can generate a bunch of types of graphs, including:
- bar
- barh (horizontal bars)
- pie
- line
- hist (histogram)
- scatter

Some of them require specific data formats. For example, bar charts are useful when dealing with the value_counts series for a given column.

```python
df.value_counts('department').nlargest(20).plot(kind = 'barh') # adding .nlargest(20) will only keep the 20 most common/largest items
```

```python
df.value_counts('culture').nlargest(20).plot(kind = 'pie') # note that the largest 20 might not be best for a pie chart because it makes it look like the 20 describe the whole sample~
```

```python
df.value_counts('artistDisplayName').nlargest(10).plot(kind = 'barh', ylabel='artist name', title='works with “woman” in the title')
```

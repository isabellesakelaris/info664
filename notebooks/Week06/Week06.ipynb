## APIs and JSON

JSON is a file that comes in the form of a dictionary or list of dictionaries. APIs send us data in JSON.

The root is the top level of the JSON file, which may have children and grand children files.

We read json files the same way we read csvs, but they are naturally interpreted as lists of dictionaries.

```python
import json #import json functionalites
with open("../notebooks/Week06/unesco_cultural_sites.json", "r", encoding="utf-8") as file:
    sites = json.load(file)
print(len(sites))
print(type(sites[0])) #shows entry 0 is a dictionary
print(sites[0].keys()) #shows all the keys
print(sites[0]['name_en']) #the english name of the first item in the sites list of dictionaries!
```

### APIs
APIs are used to query data providers in real time by grabbing data from a server. The query is provided through a url, and the response is usually formatted as a JSON object. We will look at two APIs today; the Met's API and the NYT API.

The Met API
- returns object IDs for search terms
- returns details of a given object based on object id

The NYT API 
- returns lists of articles based on search terms (does not give you the article contents)
- metadata of articles based on search terms

```python
import requests #brings in functions for getting data from the web and dealing with the responses

#the following variables make up the components of the url we'll use to query the met api
base_url = "https://collectionapi.metmuseum.org" #always the same for each met endpoint
search_path = "/public/collection/v1.1/search" #the specific endpoint for searching for object ids
search_query = "?q=Antelope" #the params for our search. In this case, q means search term and antelope is the keyword.
date_range = "&dateBegin=1900&dateEnd=2000"

#doing a search using an fstring to combine all of the query components
url = f'{base_url}{search_path}{search_query}{date_range}'
search_response = requests.get(url)

print(type(search_response))
print(dir(search_response)) #tells us all the methods and data inside search_response
```

```python
print(search_response.ok) #did it work?
```

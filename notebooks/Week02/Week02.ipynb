### Data Types in Python ###

There are a few different data types in Pythonn, including int (whole numbers), float(non-whole numbers), string (text characters), and bool (true/false).

### Variable Declaration ###
- Variables are keywords that represent other data
- Variables can be declared using the = symbol.
- The word on the left is the name of the variable, and its value is on the right.
- Anywhere below the line where the variable is assigned will remember the variable's value.

```python
name = "Isabelle" #assigning the variable 'name' as the string 'Isabelle'
print(name) #will show my name!
```

```python
age = 29 #assigning the variable 'age' as the integer 29
print(age) #this will show my age!
```

### Naming Conventions for Variables ###
- Give your variable a name that describes its value
- names **can't** start with a number
- names **can't** have a hyphen in them
- names **can** have underscores in them. We put underscores where a space would go in natural language
- names **can't** use "if" and "for", because those are special words in Python
- uppercase designates a constant variable that will not change after it is set—this is a stylistic thing

### Converting Data Types ###
- Data can be converted from one type to another using conversion functions!
- It can be really helpful if you're comparing data
- A lot of data comes as raw text

```python
pi = 3.14159
print(pi)
pi_as_string = str(pi) #str() converts to a string
print(type(pi_as_string)) #this will confirm the new variable is a string
pi_as_float = float(pi_as_string)
print(pi_as_float)
print(type(pi_as_float))

#print(pi_as_string>3) #here I get an error sayign you can't compare a string to a number
print(pi_as_float>3) #here it will work
```

### Combining and Concatenating ###
Strings can be comgined with the plus sign.
- Strings can only be concatenated with strings
- F-strings are good for formatting strings

```python
dog_name = "Duke"
dog_age = 9
# below we need to change the int to a string here (like combines with like)
my_dog = "My dog's name is " + dog_name + " and he is " + str(dog_age)
print(my_dog)
```

```python
# with F strings, you don't need to convert the int to a string beforehand
my_dog = f"My dog's name is {dog_name}. He is {dog_age}."
print(my_dog)
```

### Converting String Cases ###
- This is useful for when words appear in uppercase and lowercase across a dataset
- Here we will see a string "method," which is function dealing with a specific data type or class.
- There are built in string methods fro converting to upper, lower, capitalized, and title case

```python
print(dog_name.upper())
print(dog_name.lower())
print(my_dog.title()) #not real title case in the stylistic sense!
print(my_dog.capitalize()) #capitalizes only the first letter of the string
```

**Note:** These methods don't need parameters, but you still need the parentheses.

### Searching, Extracting, and Counting in a String ###
- Some string methods deal with searching for substrings, etc.
- The keyword **in** will evaluate true/false if something is contained within a string.

```python
print("cat" in my_dog) #will print false, since "cat" is not in the string.
print("dog" in my_dog) #will print true, since "dog" is in the string.
```

```python
my_dog = my_dog + " I really love dogs and dogs are my best friends"
print(my_dog)
```

```python
dog_count = my_dog.count("dog")
print(dog_count) #will print three
print(f"The word 'dog' appears {dog_count} times.") #will print the whole sentence!
```

```python
cat_count = my_dog.count("cat")
print(f"The word 'cat' appears {cat_count} times.")
```

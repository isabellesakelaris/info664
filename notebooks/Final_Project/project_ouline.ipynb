## Conclusion (With Assumptions and Constraints)
In the conclusion, I will (re)state what I learned during this process. For example, that while code can save time and find the "best" scanned parsing of a poem, it can't take human reasoning into account. That is, it can programmatically compare stress and position, but it can't tell us **why** Gerard Manely Hopkins might have made the metrical decisions he did, which means it favors mathematically correct parsings over artistic ones (or may miss them entirely, not taking white space and other poetic interventions into account). I will note that code is a powerful tool for analyzing poems at scale, but that a close reading approach is still valuable for a more nuanced understanding of a poet's work. Or, that code is great for learning some aspects of a poet's career (such as the types of meter he favored over time), but not all.

I will add that I used someone else's Python library (Prosodic) for this, which draws from the Carnegie Mellon University pronunciation dictionary, which is a contemporary resource and therefore isn't as applicable as a similar resource for the Queen's English during Gerard Manley Hopkins's lifetime would have been.

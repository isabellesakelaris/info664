## Review
First, bring in the moma_artists.csv dataset

```python
import csv #brings in the csv-related functions

#the open function takes three arguments: the file name, "r", and "encoding"...
#argument "r" means read the file (not write)
#argument "encoding=“utf-8”" allows all the characters to be rendered properly in utf-8

with open("moma_artists.csv", "r", encoding="utf-8") as file: # "as" creates a shorthand for file
    reader = csv.DictReader(file) #read the file as dictionary object
    data = list(reader) #converts reader from an object into a list of dictionary items
```

## String searching and filtering
Now we will do a simple search of the dataset by first name.

```python
artist_name = data[0]["DisplayName"]
print(artist_name)
```

```python
first_name_example = artist_name.split(" ") #will return a list of both, since we know the space is between two names
print(first_name_example)
first_name = artist_name.split(" ")[0] #takes the first element from the list created by the split!
print(first_name)
```

How many artists are named Mary? Make a list.

```python
# first, make an empty list to hold the artists with the name Mary
artists_named_mary = []

# then create a for loop that goes through every name
for artist in data:
    #extract each artist's first name
    first_names = artist["DisplayName"].split(" ")[0]
    #check if first_names matches "Mary" and add the matching ones to the original list
    if first_names == "Mary":
        artists_named_mary.append(artist) #it's better to add the full item if you don't know what you're going ot do with the data\
print(len(artists_named_mary))
```

You can also do this to see if Mary is in an artist's name, but not exactly Mary. That would be:
`if "Mary" in first_names` instead of `if first_names == "Mary"`

## Sorting
Now we will sort the dataset by birth year.

```python
# make copies when you modify the dataset
artists_by_year = sorted(data, key=lambda artist: int(artist["BeginDate"]))

#sorted takes two params:
# first is the dataset ("data" in this case)
# then is the sorting logic, represented by a lambda function
# here, the lmbda function is looking for each artist, then etracting BeginDate and converting it to an int
```

We need to filter out entries that have year == 0

```python
artists_with_year = []

# != means "not equal to"
for artist in data:
    if artist["BeginDate"] != "0":
        artists_with_year.append(artist)

artists_by_year_filtered = sorted(artists_with_year, key=lambda artist: int(artist["BeginDate"]))
print(artists_by_year_filtered[0])
```

```python
present_years = [] #making an empty list where I will put all the non-0 years in the dataset
for artist in data:
    if artist["BeginDate"] != "0":
        present_years.append(artist["BeginDate"])
```

# Lists

A list is a comma-separated sequence of values. The values must all be the same type of thing. We can extract diff items in a list by querying the variable name of the list plus an index number/range. The order in a list is stable, unless we modify it by sorting or grouping. Lists can also have sub-lists, which are opened and closed with brackets.

```python
print(birds)
```

You can access data in a list using its index number.

```python
print(birds[0]) #first bird
print(birds[0:3]) #first-fourth birds in a sub-list
```

If we want to access a value within a list of lists, we must use two index numbers

```python
print(bands[0]) #the first sub-list in the larger list
print(bands[0][-1]) #the last item in the first sub-list within the larger list (Ringo)
```

If we have the value, we can search for the index number!

```python
rolling_stones = bands[1]
print(rolling_stones)
'Mick Jagger' in rolling_stones #this will be true if the value is in the list
print(rolling_stones.index('Mick Jagger')) #will tell me the index numebr where 'Mick Jagger is situated!
```

I can reassign the value of any element in a list by redefining the var with its index number.

```python
print(rolling_stones[3]) #will show ronnie wood
rolling_stones[3] = "Brian Jones"
print(rolling_stones[3]) #will show brian jones
```

## Dictionaries ##
Dictionaries are like labeled data. JSON data is nested data with key:value pairs. Each attribute is separated by a comma. Each individual key and value can be a list, or another dictionary! It is a good way to organize data that isn't consistent. For example, multiple albums can have different numbers of songs on them, but we can use dictionaries to accommodate that.

While lists are ordered sets of values that can be accessed based on inde numbers, dictionareis can look up values based on keys. Dictionaries are collections of key:value pairs.

```python
beatles_albums = [
    {
        "album_title": "Please Please Me",
        "release_date": "22 March 1963",
        "running_time": "31:59",
        "tracks": ["I Saw Her Standing There", "Misery", "Anna (Go to Him)",
                   "Chains", "Boys", "Ask Me Why", "Please Please Me",
                   "Love Me Do", "P.S. I Love You", "Baby It's You",
                   "Do You Want to Know a Secret", "A Taste of Honey",
                   "There's a Place", "Twist and Shout"],
    },
    {
        "album_title": "With the Beatles",
        "release_date": "22 November 1963",
        "running_time": "33:07",
        "tracks": ["It Won't Be Long", "All I've Got to Do", "All My Loving",
                   "Don't Bother Me", "Little Child", "Till There Was You",
                   "Please Mister Postman", "Roll Over Beethoven", "Hold Me Tight",
                   "You Really Got a Hold on Me", "I Wanna Be Your Man",
                   "Devil in Her Heart", "Not a Second Time",
                   "Money (That's What I Want)"],
    },
    {
        "album_title": "A Hard Day's Night",
        "release_date": "10 July 1964",
        "running_time": "30:09",
        "tracks": ["A Hard Day's Night", "I Should Have Known Better", "If I Fell",
                   "I'm Happy Just to Dance with You", "And I Love Her", "Tell Me Why",
                   "Can't Buy Me Love", "Any Time at All", "I'll Cry Instead",
                   "Things We Said Today", "When I Get Home", "You Can't Do That",
                   "I'll Be Back"],
    }
]
#this is still a list, but it's a list of dictionaries.
```

### Adding new keys and values to a dictionary
Address a new key that doesn't yet exist and assign data to it! First we'll look at the running_time value, extract a length in seconds with some math operations, and then creating a new key in each dictionary to represent teh running time in seconds as an integer. Note that int will be more useful because it has the concept of quantity/numeric value.

```python
print(beatles_albums[0]["running_time"])
minutes_string, seconds_string = beatles_albums[0]["running_time"].split(":") #both of these will still be strings
print(minutes_string) 
print(seconds_string)
```

```python
total_seconds = int(minutes_string) * 60 + int(seconds_string) #converting minutes to seconds and adding the extra minutes
print(total_seconds)
```

```python
beatles_albums[0]["running_time_seconds"] = total_seconds #creates a new key with the value total_seconds
print(beatles_albums[0])
```

### Sorting dictionaries
When we are soritng lists, we can use a simple sort method that puts items in alphabetical or numerical order. This is more difficult when dealing with dictionaries, which have more diverse nested data. It's often the case that we don't only want to alphabetize, but sort based on a specific nested quantity.

```python
print(central_park_birds) #has a 'count' key with an int value
```

A lambda function is a small function that doesn't have a defined name. 
`x = lambda a : a + 10` will take a number a and add 10 to it. 
`print(x(5))` will return 15.

```python
#sorted() funciton takes two arguments: one is the dataset to sort, the other is the logic by which to sort.
#in this case, we use a lambda function to specify which key to sort by
#by default it goes from low > high, so passing reverse=True will make it high > low
sorted_central_park_birds = sorted(central_park_birds, key=lambda bird:bird["count"]) #inside each dictionary item, pull out the key "count" and sort according to it
print(sorted_central_park_birds)
```

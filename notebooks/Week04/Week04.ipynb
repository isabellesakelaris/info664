## Loops

```python
routine = ["wake up", "make bed", "brush teeth", "drink coffee"]
print(routine[0]) #first item in the list
```

For loop elements: 
- `for` lets the program know we are starting a loop
- `incremented variable` this can be any shorthand variable name that makes sense, often `i`

```python
grades = [56,59,1,99,55,80,90]
sum_of_grades = 0  # allows us to start from zero
for grade in grades:
    sum_of_grades = sum_of_grades + grade # adding each grade to the initial variable, which is 0 
print(sum_of_grades)
print(sum_of_grades/len(grades)) #to find the average grade
```

## Declaring Functions
We can create custom functions in python. We will use the above running time calculation as an example!


Function declarations are made of the following elements:

the def keyword: tells python that we are defining a new function!
the name of the new function(calculate_seconds)
the arguments: multiple arguments are separated by commas
the arguments are then processed by the lines of code that are indented below the "def" line
return specifies what data is output by the function

```python
# this function takes a string formatted like "30:15" and returns converts the time into an int showing the total time in seconds

def calculate_seconds(time_string): # def keyword + the function name + the arguments in parentheses + colon
    # the function is called "calculate_seconds
    # we are expecting a string formatted like "30:15" as the input
    # we will represent the input with the placeholder "time_string"
    minutes,seconds = time_string.split(":") # splitting the string into minutes and seconds
    total_seconds = int(minutes)*60 + int(seconds) # converting to ints and calculating the total seconds
    return total_seconds

print(calculate_seconds("40:45")) #example of the function using the input of 40m 45s
```

```python
# another example, for generating angry text from normal text

def angry(text):
    text = text.upper() # makes text uppercase
    exclamations = "!"*len(text) #adds as many exclamation points as there are characters in the text
    return text + exclamations

print(angry("hello world")) 
```

## Conditional Statements
conditionals rely on boolean comparison, returning `true` or `false`

- `x == y` : Are the values equal? True for yes, False for no 
- `x != y` : Are the values unequal? 
- `x < y` : Is the first value less than the second value? 
- `x > y` : Is the first value greater than the second value? 
- `x <= y` : Is the first value less than or equal to the second value?
- `x >= y` : Is the first value greater than or equal to the second value?

We can also use `and`, `or`, and `not`, aka "logic operators"

- `and` means both arguments must be true
- `or` means at least one argument must be true
- `not` means true and false are flipped

```python
x = 5
y = 6
a = 1
b = 1

print(x>y)
print(x<y)
print(x==y)
print(x<y and a!=b)
print(x>y or a==b)
```

We can also run code based on conditions using `if` and `else`. If x is true, run the code. E;se, do not run the code, or run different code.

```python
temperature = 32

if temperature < 60:
    print("put on a jacket")
else:  #else catches all data that doesn't match the condition
    print("no jacket needed!")
```

Conditional statements need the following:
- the `if` keyword, which initializes the conditional statement
- the comparison (temperature < 60), which is evaluated as True or False
- if the comparison evaluates to True, the indented portion directly below the if statement runs
- if the comparison is not True, the indented portion below else: runs

note that `false` is different from "not true"

we can also use `elif` if we want to have multiple outcomes depending on the conditions

```python
temperature = 30
if temperature >= 80: print("wear shorts") 
elif temperature >= 65: print("wear pants") 
elif temperature >= 50: print("wear a sweater") 
elif temperature >=35: print("wear a jacket") 
else: print("stay home")
```

We can also add logic operators 

```python
temperature = 70

if temperature >= 65 and temperature <= 75:
    print("the weather is comfortable")
elif temperature < 32 or temperature > 95:
    print("the weather is extreme")
```

## Using conditionals to traverse and filter a dataset
We can combine conditional statements and for loops to filter a dataset down to specific fields.

In the following example, we will iterate through all the Beatles albums in the set, go through every track in the track list, and search for the word "love". The output will be a new list of song titles containing the word "love".

```python
for album in beatles_albums:  # this is a nested for loop, in which we iterate through a parent list and its child list
    for track in album["tracks"]:
        print(track)
```

```python
for album in beatles_albums: # for each album in the set...
    for track in album["tracks"]: # and for each track in each album...
        if "love" in track.lower(): # look for "love" in the string
            # track.lower() converts the text to lower case
            print(track)
```

```python
for album in beatles_albums: # for each album in the set...
    for track in album["tracks"]: # and for each track in each album...
        if "Lov" in track: # if "lov" is found in the string
            # track.lower() converts the text to lower case
            print(track)
```

```python
for album in beatles_albums: # for each album in the set...
    for track in album["tracks"]: # and for each track in each album...
        if " lov" in (" "+track).lower(): # rensures that there is a space before "lov", so it won't capture words with "lov" in the middle
            # track.lower() converts the text to lower case
            print(track)
```

```python
love_songs = []  # this is an empty list!

for album in beatles_albums: # for each album in the set...
    for track in album["tracks"]: # and for each track in each album...
        if " lov" in (" "+track).lower(): # returns true if "love" is found in the string
            # track.lower() converts the text to lower case
            love_songs.append(track) #adds the track to the love_songs list

print(love_songs)
print(len(love_songs))
```

## CSV Data
CSV is one of the most common dataset formats, it's basically a spreadsheet with rows and columns Each row is a new line, each column is separated by a comma CSVs usually have a "header" row that is in the first row of the document. Each subsequent row is the actual data!

To work with CSV data in python we need to import the CSV library, by typing `import csv` before attempting to use any csv functions.

We can open a file using the following syntax:

`with open(file_name, options) as file:`

```python
import csv

with open("moma_artists.csv", "r", encoding="utf-8") as file: #opens the csv with utf-8 encoding and calls it "file"
    reader = csv.reader(file)  # initializing an object, "readeR" that interprets the CSV data from "file"
   
    # the below line will extract all the headings!
    header = next(reader)  # tells the reader to read the first line and move on to the next
    rows = list(reader)    # takes every other row in the dataset and makes it into a list

print(header)
print(rows[0]) # first element in list
```

We can also generate a list of dictionaries that use the header to establish a set of shared keys, assigning values for each row based on those keys/headers. This is more readable! We do this using`DictReader(file)`

```python
with open("moma_artists.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)  # this will interpret the csv as a list of dictionaries
    data = list(reader)   # store the CSV data as a list of dictionaries we can access later

print(data[0])
```

```python
# filter by nationality: only Greek artists

greek_artists = []   # create an empty list

for artist in data:  # for each artist in my data...
    if artist["Nationality"] == "Greek":
        greek_artists.append(artist)

for artist in greek_artists[:10]: #limits the list to the first 10
    print(artist["DisplayName"])
```

```python
young_artists = []

for artist in data:
    if int(artist["BeginDate"]) > 1990 and artist["Gender"] != "":
        young_artists.append(artist)
        
for artist in young_artists[:10]:
    print(artist["DisplayName"])
```

- Read the dataset as a dictionary list
- Filter that list to only include entries that have Gender data, by creating a new variable (ie artists_with_gender_data = [])
- Filter that list into three new lists using a for loop (which goes through each artist in the list), and an if/elif conditional that appends the given artist to the respective male_artists, female_artists, and non_binary_artists lists, based on whether the gender field matches. You may also want to create an other_artists field for entries that don’t match the criteria, using the else condition.
- Produce a count for each of those lists using the len() function
- Print the results into a cleanly-formatted f-string

```python
artists_with_gender_data = []
female_artists = []
male_artists = []
nonbinary_artists = []

with open("moma_artists.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file) 
    data = list(reader) 

for artist in data:
    if artist["Gender"]!= "":
        artists_with_gender_data.append(artist)
        
for artist in artists_with_gender_data:
    if artist["Gender"] == "female":
        female_artists.append(artist)
    elif artist["Gender"] == "male":
        male_artists.append(artist)
    else:
        nonbinary_artists.append(artist)

count_female_artists = (len(female_artists))
count_male_artists = (len(male_artists))
count_nonbinary_artists = (len(nonbinary_artists))

print(f"In MoMA's collection, there are {count_female_artists} female artists, {count_male_artists} male artists, and {count_nonbinary_artists} nonbinary artists.")
```
